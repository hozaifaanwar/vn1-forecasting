```python
import numpy as np
```

```python
def vn1_score(actual, forecast):
    error = forecast - actual
    mae_part = np.sum(np.abs(error)) / np.sum(actual)
    bias_part = np.abs(np.sum(error)) / np.sum(actual)
    score=mae_part + bias_part
    return score,mae_part, bias_part


print(vn1_score(np.array([10, 10, 10]), np.array([13, 12, 13])))
```
